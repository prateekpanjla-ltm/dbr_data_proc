## Bootstrap: Unity Catalog Setup for Terraform State

This notebook provisions the infrastructure needed to store Terraform state files in a Databricks Unity Catalog Volume.

**Run this ONCE manually before using GitHub Actions.**

**Resources created:**
| Resource | Name |
| --- | --- |
| Catalog | `gharchive_terraform` |
| Schema | `gharchive_terraform.tf_state` |
| Volume | `gharchive_terraform.tf_state.files` |

**State file path:** `/Volumes/gharchive_terraform/tf_state/files/terraform-{env}.tfstate`

```python
%sql
-- Create dedicated catalog for Terraform state
CREATE CATALOG IF NOT EXISTS gharchive_terraform;

-- Create schema for state files
CREATE SCHEMA IF NOT EXISTS gharchive_terraform.tf_state;

-- Create managed volume to store terraform.tfstate
CREATE VOLUME IF NOT EXISTS gharchive_terraform.tf_state.files;
```

```python
%sql
-- Grant the service principal access to read/write state files
-- Application ID: 13713ce5-dcb8-4a71-be94-28b533c10461

GRANT USE CATALOG ON CATALOG gharchive_terraform TO `13713ce5-dcb8-4a71-be94-28b533c10461`;
GRANT USE SCHEMA ON SCHEMA gharchive_terraform.tf_state TO `13713ce5-dcb8-4a71-be94-28b533c10461`;
GRANT READ VOLUME, WRITE VOLUME ON VOLUME gharchive_terraform.tf_state.files TO `13713ce5-dcb8-4a71-be94-28b533c10461`;
```

```python
%sql
-- Verify volume exists
SHOW VOLUMES IN gharchive_terraform.tf_state;

-- Verify grants on the volume
SHOW GRANTS ON VOLUME gharchive_terraform.tf_state.files;
```
